```python
Vehicle_condition_converted= []

for i in df['Vehicle_condition']:
    if i == 0:
        Vehicle_condition_converted.append('Poor condition')
    elif i == 1:
        Vehicle_condition_converted.append('Average condition')
    else:
        Vehicle_condition_converted.append('Good condition')

df['Vehicle_condition_converted']=Vehicle_condition_converted
print(df['Vehicle_condition_converted'])
```

```python
order_minutes = (
    df['Time_Orderd'].dt.hour * 60
    + df['Time_Orderd'].dt.minute
)

pickup_minutes = (
    df['Time_Order_picked'].dt.hour * 60
    + df['Time_Order_picked'].dt.minute
)

df['Order_to_Pickup_Time'] = pickup_minutes - order_minutes

# Correct for crossing midnight
df.loc[
    df['Order_to_Pickup_Time'] < 0,
    'Order_to_Pickup_Time'
] += 1440
```

```python
df['Order_to_Pickup_Time'].unique()
```

```python
df['order_hour']= df['Time_Orderd'].dt.hour
df[['order_hour','Time_Orderd']]
```

```python
order_period= []
for i in df['order_hour']:
    if i >= 5.0 and i< 12.0:
        order_period.append('Morning')
    elif i >=12.0 and i <16.0:
        order_period.append('Noon/Afternoon')
    elif i >=16.0 and i <20.0:
        order_period.append('Evening')
    else:
        order_period.append('Night')
df['order_period']= order_period

df['order_period'] 
```
